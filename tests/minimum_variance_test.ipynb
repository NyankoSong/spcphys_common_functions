```python
import spcphys_common_functions as scf
import numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.linspace(-5, 5, 100)
data = np.array([
    np.sin(x) + 0.05 * np.random.randn(100),
    np.sin(x) + np.cos(x)*0.5 + 0.1 * np.random.randn(100),
    np.cos(x)*0.5 + 0.05 * np.random.randn(100)
]).T
```

```python
fig, ax = plt.subplots(1, 4, sharex=True, sharey=True, figsize=(12, 3))

ax[0].plot(x, data[:, 0])
ax[1].plot(x, data[:, 1])
ax[2].plot(x, data[:, 2])
ax[3].plot(x, np.linalg.norm(data, axis=1))

fig.tight_layout()

ax[0].set_title('Component 1')
ax[1].set_title('Component 2')
ax[2].set_title('Component 3')
ax[3].set_title('Vector Magnitude')
```

```python
vrot, v, w = scf.minimum_variance.min_var(data, verbose=True)

fig, ax = plt.subplots(1, 4, sharex=True, sharey=True, figsize=(12, 3))
ax[0].plot(x, vrot[:, 0])
ax[1].plot(x, vrot[:, 1])
ax[2].plot(x, vrot[:, 2])
ax[3].plot(x, np.linalg.norm(vrot, axis=1))

fig.tight_layout()

ax[0].set_title('maximum variance component')
ax[1].set_title('intermediate variance component')
ax[2].set_title('minimum variance component')
ax[3].set_title('Vector Magnitude')
```

```python
fig, ax = plt.subplots(1, 1, figsize=(10, 3))
lc = scf.plot_tools.plot_colored_line(ax, vrot[:, 0], vrot[:, 1], c=x, cmap='plasma', lw=2)
ax.set_title('MV frame colored by x')
ax.set_xlabel('maximum variance component')
ax.set_ylabel('intermediate variance component')
ax.set_aspect('equal')
ax.grid(True)

plt.colorbar(lc, ax=ax, orientation='vertical', fraction=0.05, pad=0.03, aspect=10).set_label('x')
```
